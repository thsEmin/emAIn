# Part B: a tiny GPT

This notebook creates a very small model so its inputs and outputs are easy to inspect. The graded implementation is in `model.py`.

```python
import sys
from pathlib import Path
import torch

project_root = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(project_root))
from model import GPT, GPTConfig
```

```python
config = GPTConfig(vocab_size=32, block_size=8, n_layer=2, n_head=2, n_embd=8, dropout=0.0)
model = GPT(config)
tokens = torch.tensor([[1, 2, 3, 4]])
logits, loss = model(tokens, tokens)
print('Input shape:', tokens.shape)
print('Logit shape:', logits.shape)
print('Loss:', loss.item())
print('Parameters:', model.num_params())
```

## Data flow

1. Token IDs become token vectors.
2. Position vectors tell the model where tokens occur.
3. Each block applies causal attention and an MLP.
4. The causal mask prevents a token from reading future positions.
5. The final linear layer produces one score per vocabulary item.
6. Cross-entropy compares those scores with the next-token targets.
