# Part A: byte-level BPE tokenizer

This notebook uses a tiny sentence so every step is easy to inspect. The graded implementation is in `tokenizer.py`.

```python
import sys
from pathlib import Path

project_root = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(project_root))

from tokenizer import BPETokenizer
```

```python
text = 'to be or not to be'
list(text.encode('utf-8'))
```

```python
tokenizer = BPETokenizer()
tokenizer.train(text, vocab_size=264)
tokenizer.merges
```

```python
token_ids = tokenizer.encode(text)
print('Token IDs:', token_ids)
print('Decoded:', tokenizer.decode(token_ids))
assert tokenizer.decode(token_ids) == text
```

## What to explain

- Text becomes UTF-8 bytes, so IDs 0 through 255 already have a meaning.
- Training repeatedly counts neighboring pairs.
- The most frequent pair becomes a new token.
- Equal counts are resolved by choosing the smaller pair.
- Encoding repeats the learned merges in their original order.
- Decoding expands merged tokens back to bytes and then UTF-8 text.
